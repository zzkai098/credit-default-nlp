```python
import gzip
import shutil
from pathlib import Path
```

```python
input_path = Path("../data/raw/accepted_2007_to_2018Q4.csv.gz")
output_path = input_path.with_suffix("")   # -> accepted_2007_to_2018Q4.csv

with gzip.open(input_path, "rb") as f_in:
    with open(output_path, "wb") as f_out:
        shutil.copyfileobj(f_in, f_out)

print(f"Done: {output_path}")
```
