# Steps 8-9: Model Evaluation & Visualization

Compare baseline vs hybrid models using AUC-ROC, Gini, KS, and Lift analysis.

```python
import sys
sys.path.insert(0, '/home/jovyan/work/src')

import pandas as pd
import matplotlib.pyplot as plt
from evaluation import (compute_auc_roc, compute_gini, compute_ks_statistic,
                        compute_lift_table, plot_roc_curves, plot_ks_chart,
                        plot_lift_chart, plot_feature_importance, summary_comparison_table)
from models import get_model_coefficients
```

## 6.1 Load Predictions

```python
results = pd.read_csv('/home/jovyan/work/outputs/tables/model_predictions.csv')
y_true = results['y_true']
y_pred_baseline = results['pd_baseline']
y_pred_hybrid = results['pd_hybrid']
```

## 6.2 Compute Metrics

```python
auc_baseline = compute_auc_roc(y_true, y_pred_baseline)
auc_hybrid = compute_auc_roc(y_true, y_pred_hybrid)

baseline_metrics = {
    'AUC-ROC': auc_baseline,
    'Gini': compute_gini(auc_baseline),
    'KS': compute_ks_statistic(y_true, y_pred_baseline)[0],
}

hybrid_metrics = {
    'AUC-ROC': auc_hybrid,
    'Gini': compute_gini(auc_hybrid),
    'KS': compute_ks_statistic(y_true, y_pred_hybrid)[0],
}

comparison = summary_comparison_table(baseline_metrics, hybrid_metrics)
comparison
```

```python
comparison.to_csv('/home/jovyan/work/outputs/tables/metrics_comparison.csv', index=False)
```

## 6.3 ROC Curve Comparison

```python
plot_roc_curves(
    {'Baseline': (y_true, y_pred_baseline), 'Hybrid': (y_true, y_pred_hybrid)},
    save_path='/home/jovyan/work/outputs/figures/roc_comparison.png'
)
plt.show()
```

## 6.4 KS Charts

```python
plot_ks_chart(y_true, y_pred_baseline, 'Baseline KS',
             save_path='/home/jovyan/work/outputs/figures/ks_baseline.png')
plt.show()

plot_ks_chart(y_true, y_pred_hybrid, 'Hybrid KS',
             save_path='/home/jovyan/work/outputs/figures/ks_hybrid.png')
plt.show()
```

## 6.5 Lift Analysis

```python
lift_baseline = compute_lift_table(y_true, y_pred_baseline)
lift_hybrid = compute_lift_table(y_true, y_pred_hybrid)

print('Baseline Lift Table:')
display(lift_baseline)

print('Hybrid Lift Table:')
display(lift_hybrid)
```

```python
plot_lift_chart(lift_baseline, save_path='/home/jovyan/work/outputs/figures/lift_baseline.png')
plt.show()

plot_lift_chart(lift_hybrid, save_path='/home/jovyan/work/outputs/figures/lift_hybrid.png')
plt.show()
```
